```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Memuat dataset Telco Customer Churn
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

# 2. Menampilkan 5 baris pertama data
print("--- 5 Baris Pertama Data ---")
print(df[['customerID', 'tenure', 'Contract', 'MonthlyCharges', 'Churn']].head())

# 3. Visualisasi distribusi variabel target (Churn)
plt.figure(figsize=(6, 4))
sns.countplot(x='Churn', data=df, palette='Set2')
plt.title('Distribusi Kelas Pelanggan Churn (Yes vs No)')
plt.xlabel('Status Churn')
plt.ylabel('Jumlah Pelanggan')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
